episode given so we extract this - feature extraction

```python
def extract_episode(txt):
    check = False
    data = ""
    for i in txt:
        if i == ')':
            check = False
            return data
        if check == True:
            data += i
        if i == '(':
            check = True
```

```python
df['episodes'] = df['Title'].apply(extract_episode)
```

```python
df.head()
```

```python
df['episodes'] = df['episodes'].str.replace(' eps','')
```

> Make a new column for time stamp

```python
df.head()
```

```python
df.loc[2]['Title']
```

```python
def extract_time(txt):
    check = False
    data = ""
    for i in range(len(txt)):
        if txt[i] == ')': 
            for j in range(i+1,i+20):
                data += txt[j]
            return data
```

```python
df['Total Time'] = df['Title'].apply(extract_time)
```

```python
df.head()
```

Extracting months from Total Time attribute. 

```python
df.head()
```

```python
import pandas as pd

# 1. Split the 'Total Time' column into 'Start' and 'End' parts
dates = df['Total Time'].str.split(' - ', expand=True)

# 2. Convert both parts into proper datetime format
start_date = pd.to_datetime(dates[0], format='%b %Y')
end_date = pd.to_datetime(dates[1], format='%b %Y')

# 3. Calculate the difference in months
# (Years difference * 12) + Months difference + 1 (to include the starting month)
df['Months'] = (end_date.dt.year - start_date.dt.year) * 12 + (end_date.dt.month - start_date.dt.month) + 1

# View the result
print(df[['Total Time', 'Months']].head())
```
