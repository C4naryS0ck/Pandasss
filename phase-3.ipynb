```python
df
```

```python
df.fillna(df.mean())
```

# MERGING, JOINING & CONCATENATION

### MERGING 2 DF's

```python
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries)
```

```python
pd.merge(employees,salaries, on='employee_id') 
# on means common between both uske basis pr merge.
```

```python
pd.merge(employees,salaries, on='employee_id',how='inner')
# andar side me emp_id hai uske side pe merge.
# usually we use this. 
```

```python
pd.merge(employees,salaries, on='employee_id',how='outer')
```

```python
pd.merge(employees,salaries, on='employee_id',how='left')
# merge on the basis of left i.e employees. 
```

```python
pd.merge(employees,salaries, on='employee_id',how='right')
# based on salaries 
```

### CONCATENATION OF 2 DF's

```python
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
df1
```

```python
df2
```

```python
pd.concat(df1,df2)
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df2,df1])
# based on columns 
```

```python
pd.concat([df1,df2], axis=1) 
# done on the basis of rows.
```

```python
df1
```

> We have two series that we want to join.

```python
df1.join(df2)
# inner join
```

```python
df1.join(df2,how='outer')
```

```python
df2.join(df1)
```

# GROUPBY & AGGREGATION

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}

df = pd.DataFrame(data)
df
```

> category and store are categorical variable.

> Groupby category and calculate the sum of Sales. 

```python
df.groupby('Category')['Sales'].sum()
```

>  Group by Store & calculate the sum of Sales.

```python
df.groupby('Store')['Sales'].sum()
```

> Group by multiple columns.

```python
df.groupby(['Category','Store'])['Sales'].sum()
```

### AGGREGATION

It mean we want to combine the values and find something ex mean. 

```python
df
```

```python
df['Sales'].mean()
"""
mean, median, mode, min, max, count, std
"""
```

```python
df['Sales'].agg(['mean','min','max','std'])
# problem in mode. 
```
